```python
# ============================================================
# ASSIGNMENT 3 — GEMINI-ASSISTED NOTEBOOK
# 24ADC001 — BUSINESS INTELLIGENCE FOR DECISION MAKING
# Clean, Transform, Analyze
# ============================================================

# ============================================================
# CELL 1 — IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

print("Libraries imported successfully!")


# ============================================================
# CELL 2 — UPLOAD SALES DATASET
# ============================================================

print("Upload your sales CSV or Excel file.")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

print("Uploaded file:", file_name)

# Read CSV or Excel automatically
if file_name.lower().endswith(".csv"):
    df = pd.read_csv(file_name)
elif file_name.lower().endswith((".xlsx", ".xls")):
    df = pd.read_excel(file_name)
else:
    raise ValueError("Please upload a CSV or Excel file.")

print("\nDataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())


# ============================================================
# CELL 3 — STANDARDIZE COLUMN NAMES
# ============================================================

# Remove extra spaces and standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

print("Columns after standardization:")
print(df.columns.tolist())


# ============================================================
# CELL 4 — CHECK REQUIRED COLUMNS
# ============================================================

required_columns = [
    "Order_Date",
    "Region",
    "State",
    "City",
    "Category",
    "Product",
    "Sales_Amount",
    "Quantity"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    print("WARNING — These columns are missing:")
    print(missing_columns)
    print("\nYour available columns are:")
    print(df.columns.tolist())
else:
    print("All required sales columns are available!")


# ============================================================
# CELL 5 — CREATE A MESSY VERSION
# ============================================================
# Assignment requirement:
# Introduce missing values, an obvious outlier,
# and inconsistent date formatting if the dataset is clean.
#
# If your original dataset is ALREADY messy, change:
# MAKE_MESSY = False

MAKE_MESSY = True

messy_df = df.copy()

if MAKE_MESSY:

    np.random.seed(42)

    # --------------------------------------------------------
    # Introduce missing Sales_Amount
    # --------------------------------------------------------
    if len(messy_df) >= 10:
        missing_sales_index = messy_df.sample(
            n=min(5, len(messy_df)),
            random_state=42
        ).index

        messy_df.loc[missing_sales_index, "Sales_Amount"] = np.nan

    # --------------------------------------------------------
    # Introduce missing Category
    # --------------------------------------------------------
    if len(messy_df) >= 20:
        missing_category_index = messy_df.sample(
            n=min(3, len(messy_df)),
            random_state=10
        ).index

        messy_df.loc[missing_category_index, "Category"] = np.nan

    # --------------------------------------------------------
    # Introduce missing Quantity
    # --------------------------------------------------------
    if len(messy_df) >= 20:
        missing_quantity_index = messy_df.sample(
            n=min(3, len(messy_df)),
            random_state=20
        ).index

        messy_df.loc[missing_quantity_index, "Quantity"] = np.nan

    # --------------------------------------------------------
    # Introduce an obvious Sales_Amount outlier
    # --------------------------------------------------------
    if len(messy_df) > 0:
        outlier_index = messy_df["Sales_Amount"].first_valid_index()

        if outlier_index is not None:
            messy_df.loc[outlier_index, "Sales_Amount"] = (
                messy_df["Sales_Amount"].median() * 50
            )

    # --------------------------------------------------------
    # Introduce inconsistent date formatting
    # --------------------------------------------------------
    if "Order_Date" in messy_df.columns and len(messy_df) >= 10:

        date_indices = messy_df.sample(
            n=min(5, len(messy_df)),
            random_state=30
        ).index

        original_dates = pd.to_datetime(
            messy_df.loc[date_indices, "Order_Date"],
            errors="coerce"
        )

        for idx, date_value in original_dates.items():

            if pd.notna(date_value):

                # Different date formats
                formats = [
                    date_value.strftime("%d-%m-%Y"),
                    date_value.strftime("%m/%d/%Y"),
                    date_value.strftime("%Y/%m/%d")
                ]

                messy_df.loc[idx, "Order_Date"] = np.random.choice(formats)

print("Messy dataset created!")

display(messy_df.head(10))


# ============================================================
# CELL 6 — INSPECT THE MESSY DATASET
# ============================================================

print("Dataset information:")
messy_df.info()

print("\nMissing values:")
print(messy_df.isnull().sum())

print("\nNumber of duplicate rows:")
print(messy_df.duplicated().sum())


# ============================================================
# CELL 7 — REMOVE DUPLICATE RECORDS
# ============================================================

before_duplicates = len(messy_df)

messy_df = messy_df.drop_duplicates()

after_duplicates = len(messy_df)

print("Duplicate rows removed:", before_duplicates - after_duplicates)
print("Rows remaining:", len(messy_df))


# ============================================================
# CELL 8 — STANDARDIZE DATE FORMAT
# ============================================================

# Convert inconsistent date formats into one standard date format

messy_df["Order_Date"] = pd.to_datetime(
    messy_df["Order_Date"],
    errors="coerce",
    format="mixed"
)

print("Date format standardized.")

print("\nDate range:")
print("Minimum date:", messy_df["Order_Date"].min())
print("Maximum date:", messy_df["Order_Date"].max())


# ============================================================
# CELL 9 — HANDLE INVALID DATES
# ============================================================

invalid_dates = messy_df["Order_Date"].isna().sum()

print("Invalid/missing dates:", invalid_dates)

if invalid_dates > 0:

    # Fill invalid dates using the median date
    valid_dates = messy_df["Order_Date"].dropna()

    if len(valid_dates) > 0:
        median_date = valid_dates.sort_values().iloc[len(valid_dates)//2]
        messy_df["Order_Date"] = messy_df["Order_Date"].fillna(
            median_date
        )

print("Invalid dates handled.")


# ============================================================
# CELL 10 — CONVERT NUMERIC COLUMNS
# ============================================================

messy_df["Sales_Amount"] = pd.to_numeric(
    messy_df["Sales_Amount"],
    errors="coerce"
)

messy_df["Quantity"] = pd.to_numeric(
    messy_df["Quantity"],
    errors="coerce"
)

print("Numeric columns converted successfully.")


# ============================================================
# CELL 11 — HANDLE MISSING NUMERIC VALUES
# ============================================================

# Sales_Amount → median
sales_median = messy_df["Sales_Amount"].median()

messy_df["Sales_Amount"] = messy_df["Sales_Amount"].fillna(
    sales_median
)

# Quantity → median
quantity_median = messy_df["Quantity"].median()

messy_df["Quantity"] = messy_df["Quantity"].fillna(
    quantity_median
)

print("Missing numeric values handled using median values.")


# ============================================================
# CELL 12 — HANDLE MISSING CATEGORICAL VALUES
# ============================================================

categorical_columns = [
    "Region",
    "State",
    "City",
    "Category",
    "Product"
]

for column in categorical_columns:

    if column in messy_df.columns:

        if messy_df[column].isna().sum() > 0:

            mode_value = messy_df[column].mode()

            if len(mode_value) > 0:
                messy_df[column] = messy_df[column].fillna(
                    mode_value[0]
                )

print("Missing categorical values handled using mode values.")


# ============================================================
# CELL 13 — OUTLIER DETECTION USING IQR
# ============================================================

# Interquartile Range method

Q1 = messy_df["Sales_Amount"].quantile(0.25)
Q3 = messy_df["Sales_Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_condition = (
    (messy_df["Sales_Amount"] < lower_bound) |
    (messy_df["Sales_Amount"] > upper_bound)
)

outliers = messy_df[outlier_condition]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("\nNumber of detected outliers:", len(outliers))

display(outliers.head(10))


# ============================================================
# CELL 14 — HANDLE OUTLIERS
# ============================================================

# Cap extreme values at the IQR limits.
# This retains the rows instead of deleting them.

messy_df["Sales_Amount"] = messy_df["Sales_Amount"].clip(
    lower=lower_bound,
    upper=upper_bound
)

print("Outliers handled using IQR capping.")

print("\nSales statistics after outlier treatment:")
print(messy_df["Sales_Amount"].describe())


# ============================================================
# CELL 15 — FINAL DATA QUALITY CHECK
# ============================================================

print("FINAL DATA QUALITY CHECK")
print("=" * 50)

print("\nMissing values:")
print(messy_df.isnull().sum())

print("\nDuplicate rows:")
print(messy_df.duplicated().sum())

print("\nDataset shape:")
print(messy_df.shape)

display(messy_df.head())


# ============================================================
# CELL 16 — CREATE YEAR AND QUARTER
# ============================================================

messy_df["Year"] = messy_df["Order_Date"].dt.year

messy_df["Quarter"] = (
    messy_df["Order_Date"]
    .dt.to_period("Q")
    .astype(str)
)

# Create chronological quarter date
messy_df["Quarter_Start"] = (
    messy_df["Order_Date"]
    .dt.to_period("Q")
    .dt.start_time
)

print("Year and Quarter columns created.")

display(
    messy_df[
        [
            "Order_Date",
            "Year",
            "Quarter",
            "Quarter_Start"
        ]
    ].head()
)


# ============================================================
# CELL 17 — TRANSFORM DATA FOR BI REPORTING
# ============================================================
# Requirement:
# One row per Region × Quarter × Category
# with summed Sales_Amount.

bi_data = (
    messy_df
    .groupby(
        [
            "Region",
            "Quarter_Start",
            "Quarter",
            "Year",
            "Category"
        ],
        as_index=False
    )["Sales_Amount"]
    .sum()
)

bi_data = bi_data.sort_values(
    [
        "Region",
        "Quarter_Start",
        "Category"
    ]
)

print("BI reporting dataset created.")

print(
    "Rows:",
    bi_data.shape[0],
    "Columns:",
    bi_data.shape[1]
)

display(bi_data.head(20))


# ============================================================
# CELL 18 — ANALYTIC FUNCTION 1: RANK
# ============================================================
# Rank categories by sales within each Region.

bi_data["Category_Rank_in_Region"] = (
    bi_data
    .groupby(["Region", "Quarter_Start"])["Sales_Amount"]
    .rank(
        method="dense",
        ascending=False
    )
)

print("RANK analytic function completed.")

display(
    bi_data[
        [
            "Region",
            "Quarter",
            "Category",
            "Sales_Amount",
            "Category_Rank_in_Region"
        ]
    ].head(20)
)


# ============================================================
# CELL 19 — ANALYTIC FUNCTION 2: PERIOD-OVER-PERIOD
# ============================================================
# Calculate previous quarter sales.
#
# Because the BI data contains categories, we calculate
# previous-quarter sales for each Region + Category.

bi_data = bi_data.sort_values(
    [
        "Region",
        "Category",
        "Quarter_Start"
    ]
)

bi_data["Previous_Quarter_Sales"] = (
    bi_data
    .groupby(
        ["Region", "Category"]
    )["Sales_Amount"]
    .shift(1)
)

# Calculate change from previous quarter

bi_data["Quarter_Sales_Change"] = (
    bi_data["Sales_Amount"] -
    bi_data["Previous_Quarter_Sales"]
)

# Calculate percentage change

bi_data["Quarter_Sales_Growth_Percent"] = np.where(
    bi_data["Previous_Quarter_Sales"] != 0,
    (
        bi_data["Quarter_Sales_Change"] /
        bi_data["Previous_Quarter_Sales"]
    ) * 100,
    np.nan
)

print("SHIFT analytic function completed.")

display(
    bi_data[
        [
            "Region",
            "Category",
            "Quarter",
            "Sales_Amount",
            "Previous_Quarter_Sales",
            "Quarter_Sales_Change",
            "Quarter_Sales_Growth_Percent"
        ]
    ].head(20)
)


# ============================================================
# CELL 20 — ANALYTIC FUNCTION 3: RUNNING TOTAL
# ============================================================
# Running cumulative sales for each Region + Category.

bi_data = bi_data.sort_values(
    [
        "Region",
        "Category",
        "Quarter_Start"
    ]
)

bi_data["Running_Total_Sales"] = (
    bi_data
    .groupby(
        ["Region", "Category"]
    )["Sales_Amount"]
    .cumsum()
)

print("CUMSUM analytic function completed.")

display(
    bi_data[
        [
            "Region",
            "Category",
            "Quarter",
            "Sales_Amount",
            "Running_Total_Sales"
        ]
    ].head(20)
)


# ============================================================
# CELL 21 — FINAL ANALYTIC DATASET
# ============================================================

bi_data = bi_data.sort_values(
    [
        "Region",
        "Quarter_Start",
        "Category"
    ]
)

print("FINAL ANALYTIC DATASET")

display(bi_data.head(30))

print("\nFinal columns:")
print(bi_data.columns.tolist())


# ============================================================
# CELL 22 — REGION LEVEL QUARTERLY DATA
# ============================================================
# This creates a simple Region × Quarter dataset for
# overall period comparison.

region_quarter = (
    messy_df
    .groupby(
        [
            "Region",
            "Quarter_Start",
            "Quarter"
        ],
        as_index=False
    )["Sales_Amount"]
    .sum()
)

region_quarter = region_quarter.sort_values(
    [
        "Region",
        "Quarter_Start"
    ]
)

# Previous quarter
region_quarter["Previous_Quarter_Sales"] = (
    region_quarter
    .groupby("Region")["Sales_Amount"]
    .shift(1)
)

# Difference
region_quarter["Sales_Change"] = (
    region_quarter["Sales_Amount"] -
    region_quarter["Previous_Quarter_Sales"]
)

# Running total
region_quarter["Running_Total"] = (
    region_quarter
    .groupby("Region")["Sales_Amount"]
    .cumsum()
)

print("Region-level analytic dataset:")

display(region_quarter.head(20))


# ============================================================
# CELL 23 — SIMPLE VISUALIZATION
# ============================================================

plt.figure(figsize=(12, 6))

for region in region_quarter["Region"].unique():

    temp = region_quarter[
        region_quarter["Region"] == region
    ]

    plt.plot(
        temp["Quarter_Start"],
        temp["Sales_Amount"],
        marker="o",
        label=region
    )

plt.title("Quarterly Sales by Region")
plt.xlabel("Quarter")
plt.ylabel("Sales Amount")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# CELL 24 — SAVE CLEANED DATASET
# ============================================================

cleaned_file = "cleaned_sales_data.csv"

messy_df.to_csv(
    cleaned_file,
    index=False
)

print("Cleaned dataset saved as:", cleaned_file)


# ============================================================
# CELL 25 — SAVE BI DATASET
# ============================================================

bi_file = "BI_Region_Quarter_Category.csv"

bi_data.to_csv(
    bi_file,
    index=False
)

print("BI dataset saved as:", bi_file)


# ============================================================
# CELL 26 — SAVE ANALYTIC DATASET
# ============================================================

analytic_file = "analytic_sales_data.csv"

region_quarter.to_csv(
    analytic_file,
    index=False
)

print("Analytic dataset saved as:", analytic_file)


# ============================================================
# CELL 27 — DOWNLOAD FILES
# ============================================================

files.download(cleaned_file)
files.download(bi_file)
files.download(analytic_file)


# ============================================================
# CELL 28 — GEMINI EXPLANATION TO PASTE INTO MARKDOWN CELL
# ============================================================

gemini_explanation = """
### Gemini Explanation of the GroupBy/Window Code

The `groupby()` and `shift()` operations are used to perform
a period-over-period sales comparison. The data is grouped by
Region and Category so that each category is compared with its
own previous quarter within the same region. The `shift(1)`
function moves the previous quarter's sales value down by one
row, allowing the current quarter to be compared with the
previous quarter.

The `groupby()` and `cumsum()` operations calculate a running
total. For each Region and Category, the sales values are
ordered chronologically by quarter and then cumulatively added.
This produces the total sales accumulated up to each quarter.

The `rank()` operation ranks categories according to their
sales amount within a region and quarter. A dense ranking method
is used so that equal sales values receive the same rank without
unnecessary gaps in the ranking sequence.
"""

print(gemini_explanation)


# ============================================================
# CELL 29 — GEMINI EXPLANATION CRITIQUE
# ============================================================

critique = """
### My Critique of Gemini's Explanation

The Gemini explanation is accurate because it correctly describes
the purpose of groupby(), shift(), cumsum(), and rank(). In
particular, the explanation correctly states that shift(1) is
used to obtain the previous quarter's value and that cumsum()
creates a cumulative running total.

One important detail I checked was the grouping level. The
period-over-period calculation is grouped by both Region and
Category. This is necessary because comparing different
categories with each other would not represent a true
previous-quarter comparison. Therefore, I kept the Region and
Category grouping in the code.

The running total is also calculated separately for each Region
and Category and the records are sorted chronologically before
using cumsum(). Therefore, the result represents a meaningful
running total rather than an arbitrary cumulative calculation.

Overall, Gemini's explanation was useful, but I reviewed the
grouping logic and confirmed that the code needed the additional
Category grouping for an accurate quarter-to-quarter comparison.
"""

print(critique)


# ============================================================
# CELL 30 — FINAL REFLECTION (150–200 WORDS)
# ============================================================

reflection = """
### Assignment 3 Reflection

For this assignment, I used a Gemini-assisted Google Colab
notebook to clean, transform, and analyze a sales dataset. Gemini
was useful for suggesting Pandas code for handling missing values,
standardizing dates, detecting outliers, and implementing
analytical functions. I reviewed each suggestion instead of
accepting the generated code blindly.

The dataset was intentionally made messy by introducing missing
values, an extreme sales value, and inconsistent date formats.
Missing numerical values were handled using median values, while
missing categorical values were replaced using the mode. The
extreme Sales_Amount value was detected using the Interquartile
Range (IQR) method and handled using value capping. Dates were
converted into a consistent datetime format.

After cleaning, I transformed the data into a BI-friendly
Region × Quarter × Category structure. I implemented three
analytic-function equivalents in Pandas: rank() for ranking
categories, shift() for previous-quarter comparison, and
cumsum() for running totals.

Gemini's suggestions were directly useful for the basic Pandas
operations, but I had to review the grouping logic. In particular,
I ensured that period-over-period calculations were grouped by
Region and Category so that categories were compared with their
own previous quarter. This review helped me understand that
AI-generated code must be checked for business logic and not only
for syntax.
"""

print(reflection)

print("\n" + "=" * 60)
print("ASSIGNMENT 3 COMPLETED")
print("=" * 60)
print("Required components completed:")
print("1. Messy dataset")
print("2. Missing value handling")
print("3. Outlier detection and handling")
print("4. Date standardization")
print("5. BI transformation")
print("6. RANK()")
print("7. SHIFT()")
print("8. CUMSUM()")
print("9. Gemini explanation")
print("10. Gemini critique")
print("11. 150–200 word reflection")
print("=" * 60)
```
